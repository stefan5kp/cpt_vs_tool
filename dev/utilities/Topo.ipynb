```python
import requests
import time

def usgs_el(lat, lon):

    url = "https://epqs.nationalmap.gov/v1/json"
    params = {
        "x": lon,
        "y": lat,
        "units": "Meters",
        "format": "json"
    }

    elevation = None
    for attempt in range(5):
        try:
            response = requests.get(url, params=params)
            response.raise_for_status()
            elevation = float(response.json()["value"])
            break

        except requests.exceptions.HTTPError as e:
            print(f"Attempt {attempt + 1}/5 failed: {e} — retrying in 10s")
            time.sleep(10)

        except (KeyError, ValueError) as e:
            print(f"Parsing error at ({lon}, {lat}): {e}")
            break
    return print(f"({lon}, {lat}) → {elevation}m")
```

```python
usgs_el(36.882205, -120.785024)
```
