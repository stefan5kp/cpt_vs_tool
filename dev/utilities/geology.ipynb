```python
import geopandas as gpd
from shapely.geometry import Point
from pathlib import Path

path = Path(
    "wills_shp"
)

WILLS = gpd.read_file(path.joinpath("VsMAPv3_Geology_20150714.shp")).to_crs(4326)

def wills2015_vs30(lat, lon):
    """Return (geologic unit, Vs30 in m/s) from Wills et al. (2015) for a WGS84 point."""
    pt = gpd.GeoDataFrame(geometry=[Point(lon, lat)], crs=4326).to_crs(WILLS.crs)
    hit = gpd.sjoin(pt, WILLS, predicate="within")
    if hit.empty:
        return None, None          # outside the map (offshore / outside CA)
    return hit["Geologic_U"].iloc[0], float(hit["Vs30_Mean"].iloc[0])
```
